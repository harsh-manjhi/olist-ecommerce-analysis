# Olist E-Commerce Analysis

## Project Goal

In this project, we wanted to take a look at how customers behave, how well sales are going, how quickly deliveries are made and how well the sellers are doing. We used the "Olist Brazilian E-Commerce" dataset to dig into these areas and find some useful business insights that could help us make better choices about how we run things and what we do.

## Key Findings



- Revenue varies each month, suggesting that customer needs are shifting.

- A few product categories were really important, bringing in a big portion of the total revenue.

- Most of our revenue came from a few dominant customer states.

- Some sellers were doing better than others, with a few making up a large part of our sales.

- About 71 orders took longer than 100 days to arrive, and these delays were mostly in certain states, sellers and product categories.

## Recommendations



1. Investigate sellers who have had some delivery delays and see how we can improve our fulfilment processes smoother.



2. Keep an eye on the product categories that often take a while to arrive.



3. Focus marketing efforts where they’re most effective, focusing on the areas where our customers are doing well, and maybe even looking into some states that could use a bit more attention.



4. Analyze the top-performing product categories to find out where we can grow and what we need to keep in stock.
